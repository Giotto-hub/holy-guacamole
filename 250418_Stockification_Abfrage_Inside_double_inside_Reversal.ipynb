```python
from twelvedata import TDClient
import pandas as pd

# Initialize client - apikey parameter is requiered
td = TDClient(apikey="88281522a9fa4f098c77f33cff29c1ec")

# Construct the necessary time series
ts = td.time_series(
    symbol="CPRX",
    #isin= "US0378331005",
    interval="1day",
    outputsize=3,
    date="yesterday+1",
    timezone="Europe/Berlin",
)
# In DataFrame umwandeln
df = ts.as_pandas()
df = df.sort_index()  # Zeitlich aufsteigend sortieren

# Sicherstellen, dass genug Daten vorhanden sind
if len(df) < 3:
    print("❌ Nicht genug Daten für Inside-Bar-Erkennung.")
else:
    # Aktuelle und vorherige Kerze extrahieren
    curr = df.iloc[-1]
    prev = df.iloc[-2]
    prev2 = df.iloc[-3]

    # Inside Bar prüfen
    if curr['high'] < prev['high'] and curr['low'] > prev['low']:
        print("✅ Letzte Kerze ist eine Inside Bar.")
    else:
        print("🔍 Keine Inside Bar bei der letzten Kerze erkannt.")

    # Double Inside Bar prüfen
    if (
            curr['high'] < prev['high'] and curr['low'] > prev['low'] and
            prev['high'] < prev2['high'] and prev['low'] > prev2['low']):
            print("✅ Letzte Kerze ist eine Double Inside Bar.")
    else:
        print("🔍 Keine Double Inside Bar bei der letzten Kerze erkannt.")        

    # Trendumkehr (Bullish Reversal)
    if (
            prev['close'] < prev['open'] and
            curr['close'] > curr['open'] and
            curr['open'] < prev['close'] and
            curr['close'] > prev['open']):
        print("✅ Letzte Kerze iste eine Bullish Reversal.")
    else:
        print("🔍 Keine Bullish Reversal bei der letzten Kerze erkannt.")  

    # Trendumkehr (Bearish Reversal)
    if (
            prev['close'] > prev['open'] and
            curr['close'] < curr['open'] and
            curr['open'] > prev['close'] and
            curr['close'] < prev['open']
        ):
        print("✅ Letzte Kerze ist ein Bearish Reversal")
    else:
        print("🔍 Keine Bearish Reversal bei der letzten Kerze erkannt.") 
   

    # Optional: Ausgabe der Werte
    print("\nLetzte 2 Kerzen:")
    print(df.tail(3)[['open', 'high', 'low', 'close']])
```
